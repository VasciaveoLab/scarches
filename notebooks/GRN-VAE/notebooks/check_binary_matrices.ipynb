```python
import pandas as pd
```

```python
dic = {}
row_names_dic = {}
with open("/home/lgolinelli/git/scarches-1/notebooks/GRN-VAE/binary_matrices/binary_matrix_summary.txt", "w") as f:
    for perc in [1, 2, 5, 10, 25, 50]:
        path = f'/home/lgolinelli/git/scarches-1/notebooks/GRN-VAE/binary_matrices/MJ005_binary_matrix_{perc}%.csv'
        df = pd.read_csv(path)
        dic[perc] = df.loc[:, 'Nsd2'].sum()
        row_names_dic[perc] = df.loc[df['Nsd2'] == 1, 'regulator'].tolist()
        Nsd2_downstream_genes = df[df['regulator'] == 'Nsd2'].iloc[:,1:].values.sum()
        # Boolean: is Nsd2 a target of itself?
        Nsd2_self_target = bool(df.loc[df['regulator'] == 'Nsd2', 'Nsd2'].values[0]) if 'Nsd2' in df.columns and 'Nsd2' in df['regulator'].values else False
        f.write(f"------Summary for {perc}% binary matrix:------\n")
        f.write(f'Number of regulators controlling Nsd2: {dic[perc]}\n')
        f.write(f"Names of regulators controlling Nsd2: {row_names_dic[perc]}\n")
        f.write(f"Number of target genes for the Nsd2 regulator: {Nsd2_downstream_genes}\n")
        f.write(f"Is Nsd2 a target of itself? {Nsd2_self_target}\n")
        f.write("\n")
```

```python
import os

output_base_dir = '/home/lgolinelli/git/scarches-1/notebooks/GRN-VAE/binary_matrices/tsv_outputs'
os.makedirs(output_base_dir, exist_ok=True)

for perc in [1, 2, 5, 10, 25, 50]:
    path = f'/home/lgolinelli/git/scarches-1/notebooks/GRN-VAE/binary_matrices/MJ005_binary_matrix_{perc}%.csv'
    df = pd.read_csv(path)
    regulators = df['regulator']
    genes = df.columns[1:]
    # Build new DataFrame: each row is regulator, columns are genes with 1s only
    tsv_data = []
    for idx, row in df.iterrows():
        present_genes = [gene for gene in genes if row[gene] == 1]
        tsv_data.append([row['regulator']] + present_genes)
    # Save as TSV
    output_dir = os.path.join(output_base_dir, f"{perc}_percent")
    os.makedirs(output_dir, exist_ok=True)
    output_path = os.path.join(output_dir, f"regulator_gene_{perc}percent.tsv")
    with open(output_path, 'w') as f:
        for row in tsv_data:
            f.write('\t'.join(row) + '\n')
    print(f"Saved: {output_path}")
```
